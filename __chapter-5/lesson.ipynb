```python
def compound_interest(principal, rate, time):
  amount = principal * (pow((1 + rate / 100), time))
  ci = amount - principal
  print("Compound interest is", f"{ci:,.2f}")

compound_interest(500_000_000, 6, 3)

```

```python
amount = [1000, 1200, 1450, 1600, 1800]
def roi_tracker(roi: List[int]) -> None:
  roi_list = []
  for i in range(len(roi) - 1):
    a_i = roi[i]
    a_i_1 = roi[i + 1]
    roi_list.append((a_i_1 - a_i) / a_i)
  return roi_list

for i in range(len(roi_tracker(amount))):
  print(f"Roi rate {i + 1} year (%)", f"{roi_tracker(amount)[i]:,.2f}")
```

```python
def present_val(fv: float, rate: float, time: float) -> float:
  pv = fv / (pow((1 + rate / 100), time))
  return pv

print(f"{present_val(100_000_000, 6, 20):,.2f}")
```

```python
def cagr(pv: float, fv: float, n: int) -> float:
  cagr = ((fv / pv) ** (1 / n) - 1) * 100
  return cagr

print(f"{cagr(100_000_000, 1_000_000_000, 3):,.2f} %")
```
