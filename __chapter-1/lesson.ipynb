```python
from decimal import Decimal

usd_amount = Decimal(input('Enter the amount in USD: '))
usd_to_eur_rate = Decimal('0.892')
usd_to_vnd_rate = Decimal('25926.50')
usd_to_jpy_rate = Decimal('144.50')
usd_to_cny_rate = Decimal('6.92')
usd_to_gbp_rate = Decimal('0.77')
usd_to_aud_rate = Decimal('1.47')
usd_to_cad_rate = Decimal('1.35')
usd_to_chf_rate = Decimal('0.92')

# eur_amount = usd_amount * usd_to_eur_rate
# vnd_amount = usd_amount * usd_to_vnd_rate
# jpy_amount = usd_amount * usd_to_jpy_rate
# cny_amount = usd_amount * usd_to_cny_rate
# gbp_amount = usd_amount * usd_to_gbp_rate
# aud_amount = usd_amount * usd_to_aud_rate
# cad_amount = usd_amount * usd_to_cad_rate
# chf_amount = usd_amount * usd_to_chf_rate

def exchange_msg(amount, rate, currency):
  print(f"${amount:,.2f} USD is equal to {amount * rate:,.2f} {currency}")

currencies = {
  'EUR': usd_to_eur_rate,
  'VND': usd_to_vnd_rate,
  'JPY': usd_to_jpy_rate,
  'CNY': usd_to_cny_rate,
  'GBP': usd_to_gbp_rate,
  'AUD': usd_to_aud_rate,
  'CAD': usd_to_cad_rate,
  'CHF': usd_to_chf_rate
}

for currency, rate in currencies.items():
    exchange_msg(usd_amount, rate, currency)

# print(f"${usd_amount:,.2f} USD is equal to {eur_amount:,.2f} EUR")
# print(f"${usd_amount:,.2f} USD is equal to {vnd_amount:,.2f} VND")
# print(f"${usd_amount:,.2f} USD is equal to {jpy_amount:,.2f} JPY")
# print(f"${usd_amount:,.2f} USD is equal to {cny_amount:,.2f} CNY")
# print(f"${usd_amount:,.2f} USD is equal to {gbp_amount:,.2f} GBP")
# print(f"${usd_amount:,.2f} USD is equal to {aud_amount:,.2f} AUD")
# print(f"${usd_amount:,.2f} USD is equal to {cad_amount:,.2f} CAD")
# print(f"${usd_amount:,.2f} USD is equal to {chf_amount:,.2f} CHF")
```

```python
from decimal import Decimal

name = input('name?')
amount = Decimal(input('amount?'))
tip_percentage = Decimal(input('tip %?'))
tip_amount = Decimal(amount * tip_percentage / 100)
total = Decimal(amount + tip_amount)

print(f'total bill: ${total:,.2f}')
```

```python
electric_bill = Decimal(input('electric bill?'))
water_bill = Decimal(input('water bill?'))
school_fee = Decimal(input('school fee?'))
extra_class_fee = Decimal(input('extra class fee?'))
gift = Decimal(input('gift?'))

total = electric_bill + water_bill + school_fee + extra_class_fee + gift

print(f'total expense: ${total:,.2f}')
```
