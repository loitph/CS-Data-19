```python
from decimal import Decimal

name = input('name?')
amount = Decimal(input('amount?'))
tip = Decimal(input('tip percen?'))
til_calc = amount * tip / 100
total = amount + til_calc

print(f'total bill: ${total}')
```

```python
electric_bill = Decimal(input('electric bill?'))
water_bill = Decimal(input('water bill?'))
school_fee = Decimal(input('school fee?'))
extra_class_fee = Decimal(input('extra class fee?'))
gift = Decimal(input('gift?'))

total = electric_bill + water_bill + school_fee + extra_class_fee + gift

print(f'total expense: ${total}')
```
