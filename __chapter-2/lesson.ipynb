```python
employee_id = input('employee id?').upper()

match employee_id:
  case 'NV001':
    print('John Doe - Sales - Salary: $1000')
  case 'NV002':
    print('Jane Smith - Marketing - Salary: $1200')
  case 'NV003':
    print('Alice Johnson - HR - Salary: $1100')
  case _:
    print('Employee not found')
```

```python
from decimal import Decimal

score = Decimal(input('Enter your score: '))

match score:
  case s if s >= 9:
    print('Grade: A')
  case s if 8 <= s < 9:
    print('Grade: B')
  case s if 6.5 <= s < 8:
    print('Grade: C')
  case s if 5 <= s < 6.5:
    print('Grade: D')
  case s if s < 6:
    print('Grade: F')
```
