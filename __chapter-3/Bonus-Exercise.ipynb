# **Bài tập 1: Hệ thống tính tổng thu nhập nhân viên cuối tháng**

### Bối cảnh:
Bạn có một danh sách nhân viên của một công ty chứa các thông tin về lương cứng, số lượng dự án họ đã hoàn thành trong tháng và tỷ lệ số ngày đi muộn. Hãy viết chương trình Python để tính toán thu nhập thực tế cuối tháng của từng người theo các quy tắc sau:

1. **Tính tiền thưởng dự án (Phép nhân & Điều kiện if):**
   * Nếu hoàn thành từ 3 dự án trở lên: Thưởng 3,000,000đ cho mỗi dự án.
   * Nếu hoàn thành từ 1 đến 2 dự án: Thưởng 1,500,000đ cho mỗi dự án.
   * Nếu không hoàn thành dự án nào: Không có tiền thưởng.

2. **Tính tiền phạt đi muộn (Phép nhân):**
   * Tiền phạt bằng `Luong_Cung` nhân với `Ty_Le_Di_Muon`.

3. **Tính tổng thu nhập thực nhận (Phép cộng & Phép trừ):**
   * `Tong_Thu_Nhap = Luong_Cung + Tien_Thuong - Tien_Phat`

4. **Phân loại hiệu suất (Điều kiện if - elif - else):**
   * Nếu số dự án hoàn thành từ 3 trở lên và tỷ lệ đi muộn dưới 0.05: Xếp loại **"Xuất sắc"**.
   * Nếu số dự án hoàn thành từ 1 trở lên: Xếp loại **"Đạt yêu cầu"**.
   * Các trường hợp còn lại: Xếp loại **"Cần cải thiện"**.

---

```python
import numpy as np
import pandas as pd


# Bonus:
# projects > 3 projects -> 3_000_000 VND/project
# 1 <= projects <= 3 projects -> 1_500_000 VND/project
# < 1 project -> 0 VND/project

# Late:
# Tien_Phat = Luong_Cung * Ty_Le_Di_Muon

# Gross:
# salary = Luong_Cung + Bonus - Tien_Phat

# Ranking:
# 3 projects and Late < 5% -> Excellent
# 1 <= projects <= 3 projects -> Good
# Others -> Need to improve

employee_data = pd.read_csv('employee.csv')

danh_sach_thuong = []
danh_sach_phat = []
danh_sach_tong_thu_nhap = []
danh_sach_xep_loai = []

for index, employee in employee_data.iterrows():
  ten = employee['Ten_Nhan_Vien']
  luong_cung = employee['Luong_Cung']
  so_du_an = employee['So_Du_An_Hoan_Thanh']
  ty_le_muon = employee['Ty_Le_Di_Muon']

  thuong = 0
  if so_du_an > 3:
    thuong = 3_000_000
  elif 1 <= so_du_an <= 3:
    thuong = 1_500_000
  else:
    thuong = 0
  danh_sach_thuong.append(f'{thuong:,.2f} VND')

  tien_phat = luong_cung * ty_le_muon
  danh_sach_phat.append(f'{tien_phat:,.2f} VND')

  danh_sach_tong_thu_nhap.append(f'{luong_cung + thuong - tien_phat:,.2f} VND')

  ranking = 'Xuất sắc'
  if so_du_an >= 3 and ty_le_muon <= 0.05:
    ranking = 'Xuất sắc'
  elif 1 <= so_du_an:
    ranking = 'Đạt yêu cầu'
  else:
    ranking = 'Cần cải thiện'

  danh_sach_xep_loai.append(ranking)

employee_data['Tien_Thuong'] = danh_sach_thuong
employee_data['Tien_Phat'] = danh_sach_phat
employee_data['Tong_Thu_Nhap'] = danh_sach_tong_thu_nhap
employee_data['Xep_Loai'] = danh_sach_xep_loai

employee_data.tail(10)

```

# **I. Đề bài thực hành: Hệ thống Quản lý và Khuyến nghị Danh mục Chứng khoán**

### Bối cảnh:
Bạn có danh sách các mã cổ phiếu mà khách hàng đang nắm giữ kèm theo giá mua vào, giá thị trường hiện tại và số lượng. Hãy viết chương trình Python để tính toán hiệu suất danh mục và đưa ra khuyến nghị hành động cho từng mã theo các quy tắc sau:

1. **Tính tổng chi phí mua vào (Phép nhân):**
   * `Chi_Phi_Mua = So_Luong * Gia_Mua`

2. **Tính giá trị hiện tại của cổ phiếu (Phép nhân):**
   * `Gia_Tri_Hien_Tai = So_Luong * Gia_Thi_Truong`

3. **Tính phần trăm lợi nhuận/thua lỗ (Phép trừ, Phép chia & Phép nhân):**
   * `Phan_Tram_Loi_Nhuan = ((Gia_Thi_Truong - Gia_Mua) / Gia_Mua) * 100`

4. **Đưa ra Khuyến nghị hành động (Điều kiện if - elif - else):**
   * Nếu phần trăm lợi nhuận lớn hơn hoặc bằng **15%**: Khuyến nghị **"CHỐT LỜI (Take Profit)"**.
   * Nếu phần trăm thua lỗ chạm hoặc vượt quá **-7%** (tức là `Phan_Tram_Loi_Nhuan <= -7`): Khuyến nghị **"CẮT LỖ (Cut Loss)"**.
   * Các trường hợp còn lại (nằm trong khoảng từ -7% đến dưới 15%): Khuyến nghị **"TIẾP TỤC NẮM GIỮ (Hold)"**.

---

```python
import pandas as pd

stock_data = pd.read_csv('stock_data.csv')

ds_chi_phi_mua = []
ds_gt_hien_tai = []
ds_loi_nhuan = []
ds_khuyen_nghi = []

for index, stock in stock_data.iterrows():
  chi_phi_mua = stock['So_Luong'] * stock['Gia_Mua']
  ds_chi_phi_mua.append(f'{chi_phi_mua:,.2f} VND')

  gia_tri_hien_tai = stock['So_Luong'] * stock['Gia_Thi_Truong']
  ds_gt_hien_tai.append(f'{gia_tri_hien_tai:,.2f} VND')

  loi_nhuan = ((stock['Gia_Thi_Truong'] - stock['Gia_Mua']) / stock['Gia_Mua']) * 100
  ds_loi_nhuan.append(f'{loi_nhuan:,.2f} %')

  khuyen_nghi = ''
  if loi_nhuan >= 15:
    khuyen_nghi = 'Take Profit'
  elif loi_nhuan <= -7:
    khuyen_nghi = 'Cut Loss'
  else:
    khuyen_nghi = 'Hold'
  ds_khuyen_nghi.append(khuyen_nghi)

stock_data['Chi_Phi_Mua'] = ds_chi_phi_mua
stock_data['Gia_Tri_Hien_tai'] = ds_gt_hien_tai
stock_data['Loi_Nhuan'] = ds_loi_nhuan
stock_data['Khuyen_Nghi'] = ds_khuyen_nghi

stock_data.tail(10)
```
