```python
n = int(input("Enter a number: "))

sum = (n * (n + 1)) / 2

print(f'1/ Using formula: {sum}')


sum = 0
for i in range(1, n+1):
  sum += i

print(f'2/ Using loop: {sum}')
```

```python
n = int(input('Enter a number: '))
pow = int(input('Enter a power: '))

print(f'1/ {n}^({pow}) = {n**pow}')


result = 1
for i in range(pow):
  result *= n

print(f'2/ loop result = {result}')
```

```python
factorial = int(input('Enter a number: '))

result = 1

if factorial == 0 or factorial == 1:
  print(result)
else:
  for i in range(1, factorial + 1):
    result *= i

  print(result)
```
