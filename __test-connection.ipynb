# Chapter 1 — PostgreSQL connection

Connection settings live in `.env`. Run the setup cell once per session, then use `%%sql` cells.

```sql
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine, URL

load_dotenv()

pg_url = URL.create(
    "postgresql+psycopg",
    username=os.environ["PG_USER"],
    password=os.environ["PG_PASSWORD"],
    host=os.environ["PG_HOST"],
    port=int(os.environ["PG_PORT"]),
    database=os.environ["PG_DATABASE"],
)
engine = create_engine(pg_url)

%load_ext sql
%config SqlMagic.displaylimit = 50
%sql engine
```

```sql
%%sql
SELECT version();
```

Query results can also be pulled into pandas:

```sql
import pandas as pd

pd.read_sql("SELECT current_database(), current_user, now()", engine)
```
