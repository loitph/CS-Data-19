```python
from datetime import datetime

current_year = datetime.now().year

participants = [
  {
    'name': 'John',
    'dob': 2003,
    'gender': 'M',
    'is_married': 'Single',
    'education': 'Bachelor',
    'stress': 80,
    'anxiety': 75,
    'depression': 70,
    'sleep': 30,
    'social': 20
  },
  {
    'name': 'Marie',
    'dob': 2001,
    'gender': 'F',
    'is_married': 'Single',
    'education': 'Bachelor',
    'stress': 30,
    'anxiety': 25,
    'depression': 10,
    'sleep': 90,
    'social': 40
  },
  {
    'name': 'Bob',
    'dob': 2000,
    'gender': 'M',
    'is_married': 'Married',
    'education': 'Bachelor',
    'stress': 40,
    'anxiety': 30,
    'depression': 20,
    'sleep': 80,
    'social': 50
  },
  {
    'name': 'Alice',
    'dob': 1999,
    'gender': 'F',
    'is_married': 'Married',
    'education': 'Bachelor',
    'stress': 50,
    'anxiety': 40,
    'depression': 30,
    'sleep': 70,
    'social': 60
  },
  {
    'name': 'Alice',
    'dob': 1999,
    'gender': 'F',
    'is_married': 'Married',
    'education': 'Bachelor',
    'stress': 50,
    'anxiety': 40,
    'depression': 30,
    'sleep': 70,
    'social': 60
  },
  {
    'name': 'Alice',
    'dob': 1999,
    'gender': 'F',
    'is_married': 'Married',
    'education': 'Master',
    'stress': 50,
    'anxiety': 40,
    'depression': 30,
    'sleep': 70,
    'social': 60
  },
  {
    'name': 'Alice',
    'dob': 1999,
    'gender': 'F',
    'is_married': 'Married',
    'education': 'Master',
    'stress': 50,
    'anxiety': 40,
    'depression': 30,
    'sleep': 70,
    'social': 60
  }
]

print("""
*** Khảo sát sức khoẻ tinh thần ***
Ex: John, 2003, M, Single, Bachelor, 80, 75, 70, 30, 20
Def: Name, DoB, Gender, isMarried, Graduate, Stress, Anxiety, Depression, Sleep, Social

""")

while(
  line := input('Input data (type exit to escape!): ')
) != 'exit':
  values = [v.strip() for v in line.split(',')]

  if len(values) != 10:
    print('Invalid input!')
    continue

  (
    name, dob, gender, is_married, education, stress, anxiety, depression, sleep, social
  ) = values

  is_duplicate = False
  for person in participants:
    if person['name'].lower() == name.lower() and person['dob'] == int(dob):
      is_duplicate = True
      print('Duplicate person!')
      # break

  if is_duplicate:
    break

  participants.append({
    'name': name,
    'dob': int(dob),
    'age': current_year - int(dob),
    'gender': gender,
    'is_married': is_married,
    'education': education,
    'stress': int(stress),
    'anxiety': int(anxiety),
    'depression': int(depression),
    'sleep': int(sleep),
    'social': int(social)
  })

if not participants:
  print('No data!')


total_age = 0
total_stress = 0
total_anxiety = 0
total_depression = 0
total_sleep = 0
total_social = 0

high_risk_count = 0

highest_risk_person = None

single_count = 0
single_risk_sum = 0

married_count = 0
married_risk_sum = 0

male_count = 0
male_stress_sum = 0

female_count = 0
female_stress_sum = 0

education_stats = {}

print('--- Details ---')
for person in participants:
  risk_index = (
    person['stress'] + person['anxiety'] + person['depression'] + person['sleep'] + person['social']
  )

  person['risk_index'] = risk_index

  total_age += person.get('age', current_year - person['dob'])
  total_stress += person['stress']
  total_anxiety += person['anxiety']
  total_depression += person['depression']
  total_sleep += person['sleep']
  total_social += person['social']

  if risk_index > 50:
    high_risk_count += 1

  if highest_risk_person is None or risk_index > highest_risk_person['risk_index']:
    highest_risk_person = person

  if person['is_married'] == 'Single':
    single_count += 1
    single_risk_sum += risk_index
  elif person['is_married'] == 'Married':
    married_count += 1
    married_risk_sum += risk_index

  if person['gender'] == 'M':
    male_count += 1
    male_stress_sum += person['stress']

  elif person['gender'] == 'F':
    female_count += 1
    female_stress_sum += person['stress']

  edu = person['education']
  if edu not in education_stats:
    education_stats[edu] = { 'count': 0, 'stress_sum': 0 }

  education_stats[edu]['count'] += 1
  education_stats[edu]['stress_sum'] += person['stress']

  age = person.get('age', current_year - person['dob'])
  print(
      f"{person['name']:<10}"
      f"Age:{age:<3}"
      f" Gender:{person['gender']:<2}"
      f" Marital:{person['is_married']:<8}"
      f" Education:{person['education']:<10}"
      f" Risk:{risk_index}"
  )

n = len(participants)
result = {
  "participants": n,
  "avg_age": total_age / n,
  "avg_stress": total_stress / n,
  "avg_anxiety": total_anxiety / n,
  "avg_depression": total_depression / n,
  "avg_sleep": total_sleep / n,
  "avg_social": total_social / n,
  "high_risk_rate": high_risk_count / n * 100,
  "highest_risk_person": highest_risk_person,
  "single_count": single_count,
  "single_risk_sum": single_risk_sum,
  "married_count": married_count,
  "married_risk_sum": married_risk_sum,
  "male_count": male_count,
  "male_stress_sum": male_stress_sum,
  "female_count": female_count,
  "female_stress_sum": female_stress_sum,
  "education_stats": education_stats
}

print("--- Result ---")
for key, value in result.items():
  # print(key, value)
  if isinstance(value, float):
    print(f"{key}: {value:,.2f}")


print("--- The highest risk person ---")
if highest_risk_person:
  print(
      f"{highest_risk_person['name']:<10}"
      f"Age:{highest_risk_person.get('age', current_year - highest_risk_person['dob']):<3}"
      f" Gender:{highest_risk_person['gender']:<2}"
      f" Marital:{highest_risk_person['is_married']:<8}"
      f" Education:{highest_risk_person['education']:<10}"
      f" Risk:{highest_risk_person['risk_index']}"
  )

print("--- Gender statistics ---")
if male_count > 0:
  print(f"Male stress avg: {male_stress_sum / male_count:,.2f}")

if female_count > 0:
  print(f"Female stress avg: {female_stress_sum / female_count:,.2f}")
```
